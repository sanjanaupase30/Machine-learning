```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

Create Dataset

```python
data={
    "size":[700,900,1000,1000,1200,1500,1500,1700,1800,1800,2000,2200],
    "bhk":[1,1,1,2,2,2,3,3,3,4,4,4],
    "price":[1800000,2000000,2200000,2500000,2750000,3200000,3500000,4000000,4200000,4500000,4800000,5100000]
}
```

```python
print(data)
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

```python
df.corr()
```

visualise using seaborn


```python
sbn.pairplot(df)
```

Dependant(y) and Independant(X) features

```python
X = df[["size","bhk"]]
print(X)
y = df["price"]
print(y)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.15,random_state=42)
```

Model Training


```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

Scatter plot

```python
plt.scatter(df["size"],df["price"])
plt.scatter(df["price"],df["bhk"])
plt.scatter(df["size"],df["bhk"])
```

view model parameters

```python
print("intercept",model.intercept_)
```

```python
print("coefficient",model.coef_)
```

Make Predictions

```python
y_pred=model.predict(X_test)
print("predicted :",y_pred)
print("actual :",y_test.values)
```

evaluate model

```python
mse=mean_squared_error(y_test,y_pred)
print(mse)
```

```python
r2=r2_score(y_test,y_pred)
print(r2)
```

Predict the price of new flat

```python
a=[[1800,2]]
predicted_price=model.predict(a)
print("predicted price is :",predicted_price)
```
